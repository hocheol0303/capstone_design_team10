```python
# 전처리 된 이미지의 경로 정리
import re
import math

id_set = set()
image_df = pd.DataFrame(columns=['user_id', 'forehead_path', 'right_eye_path', 'left_eye_path', 'nasolabial_path', 'perioral_path', 'right_vol_path', 'left_vol_path'])

image_dict = {}

# label_df를 가지고 rule base로 이미지 경로를 정리
for i, row in label_df.iterrows():
    # print(row)
    user_id = row['filename'].split('_')[0]
    
    # print(user_id)
    # print()
    for key, value in row[3:].items():
        # print(key, value)
        if math.isnan(value):
            continue
        else:
            image_path = os.path.join('030_data/data', f"{key}_{row['filename']}.jpg")
            abs_path = os.path.join(wrk_root, image_path)
            if not os.path.exists(abs_path):
                print(f"라벨은 존재하지만 이미지가 존재하지 않는다. : {abs_path}")
            else:
                # if user_id not in image_df['user_id'].values:
                #     image_df = image_df.append({'user_id': user_id, f"{key}_path": image_path}, ignore_index=True)
                # else:
                #     image_df.loc[image_df['user_id'] == user_id, f"{key}_path"] = image_path

                if user_id not in image_dict.keys():
                    image_dict[user_id] = {'user_id': user_id}
                
                image_dict[user_id][key] = image_path
```

```python
num = 0
for user_id, item in image_dict.items():
    for key, value in item.items():
        if key != 'user_id':
            num += 1

print(f"label을 통해 확인한 이미지 개수: {num}, 실제 이미지 개수: {len(os.listdir(os.path.join(wrk_root, '030_data/data')))}")

image_df = pd.DataFrame.from_dict(image_dict, orient='index')
# image_df

image_df = image_df.reset_index(drop=True)
image_df.rename(columns={col:f"{col}_path" for col in image_df.columns[1:]}, inplace=True)
image_df
```

```python
label_df['filename'] = label_df['filename'].apply(lambda x:x.split('_')[0])
label_df['성별'] = label_df['성별'].apply(lambda x:'male' if x == '남자' else 'female')

label_df.rename(columns={'filename': 'user_id', '성별': 'gender', '나이': 'age'}, inplace=True)
label_df
```

```python
merged_df = pd.merge(image_df[['user_id', 'forehead_path', 'right_eye_path', 'left_eye_path', 'nasolabial_path', 'perioral_path', 'right_vol_path', 'left_vol_path']], label_df, on='user_id')
merged_df = merged_df[['user_id', 'age', 'gender', 'forehead_path', 'right_eye_path', 'left_eye_path', 'nasolabial_path', 'perioral_path', 'right_vol_path', 'left_vol_path', 'forehead', 'right_eye', 'left_eye', 'nasolabial', 'perioral', 'right_vol', 'left_vol']]

merged_df.fillna(-1, inplace=True)
merged_df = merged_df.astype({
    'forehead': 'int', 
    'right_eye': 'int', 
    'left_eye': 'int', 
    'nasolabial': 'int', 
    'perioral': 'int', 
    'right_vol': 'int', 
    'left_vol': 'int'
})
merged_df
```

```python
import json

with open(os.path.join(wrk_root, '030_data/label/wrinkle_label.json'), 'w') as f:
    json.dump(merged_df.to_dict(orient='records'), f, indent=2, ensure_ascii=False)
```

```python
# 전처리 된 이미지의 경로 정리
import re
import math

id_set = set()
image_df = pd.DataFrame(columns=['user_id', 'forehead_path', 'right_eye_path', 'left_eye_path', 'nasolabial_path', 'perioral_path', 'right_vol_path', 'left_vol_path'])

image_dict = {}

# label_df를 가지고 rule base로 이미지 경로를 정리
for i, row in label_df.iterrows():
    # print(row)
    user_id = row['filename'].split('_')[0]
    image_id = re.search(r'[0-9]+_[0-9]+', row['filename']).group(0)
    
    for key, value in row[1:].items():
        # print(key, value)
        if math.isnan(value):
            continue
        else:
            image_path = os.path.join('028_data/data', f"{key}_{row['filename']}.jpg")
            abs_path = os.path.join(wrk_root, image_path)
            if not os.path.exists(abs_path):
                print(f"라벨은 존재하지만 이미지가 존재하지 않는다. : {abs_path}")
            else:
                if image_id not in image_dict.keys():
                    image_dict[image_id] = {'user_id': user_id, 'image_id': image_id}
                
                image_dict[image_id][key] = image_path
```

```python
num = 0
for image_id, item in image_dict.items():
    for key, value in item.items():
        if 'id' not in key:
            num += 1

print(f"label을 통해 확인한 이미지 개수: {num}, 실제 이미지 개수: {len(os.listdir(os.path.join(wrk_root, '028_data/data')))}")

image_df = pd.DataFrame.from_dict(image_dict, orient='index')
# image_df

image_df = image_df.reset_index(drop=True)
image_df.rename(columns={col:f"{col}_path" for col in image_df.columns[2:]}, inplace=True)
image_df
```

```python
import os
import shutil

src_root = '/home/work/hocheol_dir/workspace/data/034_cropped_data'
dst_root = '/home/work/hocheol_dir/workspace/data/fixed_data'

for dirpath, dirnames, filenames in os.walk(src_root):
    for filename in filenames:
        if filename.lower().endswith('.jpg'):
            src_path = os.path.join(dirpath, filename)

            # left <-> right 이름 바꾸기
            if 'left_cheek' in filename:
                new_filename = filename.replace('left_cheek', 'right_cheek')
            elif 'right_cheek' in filename:
                new_filename = filename.replace('right_cheek', 'left_cheek')
            else:
                continue  # left/right가 없으면 스킵

            # 상대 경로 유지
            relative_path = os.path.relpath(dirpath, src_root)
            dst_dir = os.path.join(dst_root, relative_path)
            os.makedirs(dst_dir, exist_ok=True)

            dst_path = os.path.join(dst_dir, new_filename)

            # 파일 복사
            shutil.copy2(src_path, dst_path)
            print(f'Copied: {src_path} -> {dst_path}')
```

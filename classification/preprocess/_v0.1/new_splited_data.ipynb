```python
import os
import json

dest_dir = "/home/work/hocheol_dir/workspace/data/034_cropped_data/0512_dataset"
origin_dir = "/home/work/hocheol_dir/workspace/data/034_cropped_data/origin"

tr_json_path = os.path.join(dest_dir, "train.json")
val_json_path = os.path.join(dest_dir, "val.json")
test_json_path = os.path.join(dest_dir, "test.json")

with open(tr_json_path, "r") as f:
    tr_json = json.load(f)
with open(val_json_path, "r") as f:
    val_json = json.load(f)
with open(test_json_path, "r") as f:
    test_json = json.load(f)

len(tr_json), len(val_json), len(test_json)
```

```python
origin_tr = set(os.listdir(os.path.join(origin_dir, '1-2.Training')))
origin_val = set(os.listdir(os.path.join(origin_dir, '2-2.Validation')))
```

```python
# origin_dir, origin_tr
```

```python
dest_tr = set([data['id'].split('_')[0] for data in tr_json])
dest_val = set([data['id'].split('_')[0] for data in val_json])
dest_test = set([data['id'].split('_')[0] for data in test_json])
print(len(dest_tr), len(dest_val), len(dest_test))
```

```python
import shutil
from tqdm import tqdm

for id in tqdm(origin_tr):
    if id in dest_tr:
        shutil.copytree(os.path.join(origin_dir, "1-2.Training", id), os.path.join(dest_dir, "Training", id), dirs_exist_ok=True)
    elif id in dest_val:
        shutil.copytree(os.path.join(origin_dir, "1-2.Training", id), os.path.join(dest_dir, "Validation", id), dirs_exist_ok=True)
    elif id in dest_test:
        shutil.copytree(os.path.join(origin_dir, "1-2.Training", id), os.path.join(dest_dir, "Test", id), dirs_exist_ok=True)

for id in tqdm(origin_val):
    if id in dest_tr:
        shutil.copytree(os.path.join(origin_dir, "2-2.Validation", id), os.path.join(dest_dir, "Training", id), dirs_exist_ok=True)
    elif id in dest_val:
        shutil.copytree(os.path.join(origin_dir, "2-2.Validation", id), os.path.join(dest_dir, "Validation", id), dirs_exist_ok=True)
    elif id in dest_test:
        shutil.copytree(os.path.join(origin_dir, "2-2.Validation", id), os.path.join(dest_dir, "Test", id), dirs_exist_ok=True)
```

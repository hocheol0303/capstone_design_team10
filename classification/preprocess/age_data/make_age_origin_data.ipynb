```python
import os
exist_origin_dir = '/home/work/hocheol_dir/datasets/_origin_data'
new_origin_dir = '/home/work/hocheol_dir/workspace/datasets2/age_data/origin_data'
dirs = [dir for dir in os.listdir(exist_origin_dir) if dir.endswith('_data') and 'acne' not in dir]
dirs.sort()
for dir in dirs:
    os.makedirs(os.path.join(new_origin_dir, dir, 'data'), exist_ok=True)
```

```python
# origin_dir 아래에 ###_data/data 에는 바로 이미지 파일들이 들어있도록. 999는 제외. path에 각 폴더명이 포함되어 있음
```

```python
# data 폴더 아래에 이미지 파일 바로 없는거 : 028, 034, 045, 092, 999
# 028 해둠. 999는 path에 그 구조가 포함되어 있어서 괜찮을듯?
```

```python
for dir in dirs:
    if '999' in dir:
        continue
    else:
        exist_data_dir = os.path.join(exist_origin_dir, dir, 'data')
        new_data_dir = os.path.join(new_origin_dir, dir, 'data')
        for root, _, files in os.walk(exist_data_dir):
            for file in files:
                if not os.path.exists(os.path.join(new_data_dir, file)):
                    os.symlink(os.path.join(root, file), os.path.join(new_data_dir, file))
```

```python
exist_data_dir = os.path.join(exist_origin_dir, '999_data', 'data')
new_data_dir = os.path.join(new_origin_dir, '999_data', 'data')
for root, _, files in os.walk(exist_data_dir):
    for file in files:
        if 'SNUH_final' in root:
            os.symlink(os.path.join(root, file), os.path.join(new_data_dir, 'SNUH_final', file))
        elif 'online' in root:
            os.symlink(os.path.join(root, file), os.path.join(new_data_dir, 'online', file))
```

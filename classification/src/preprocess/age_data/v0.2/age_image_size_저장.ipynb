```python
import cv2
import os
import json
from tqdm import tqdm

paths_json = '/home/work/hocheol_dir/workspace/datasets/v0.2_data/age/age_label_paths.json'
img_root = '/home/work/hocheol_dir/workspace/datasets/v0.2_data/age'

with open(paths_json, 'r') as f:
    paths = json.load(f)
for path in tqdm(paths):
    for step in ['train', 'val', 'test']:
        json_path = os.path.join(path, f"age_{step}.json")
        new_json_path = os.path.join(path, 'add_img_size', f"age_{step}.json")
        os.makedirs(os.path.dirname(new_json_path), exist_ok=True)
        with open(json_path, 'r') as f:
            json_data = json.load(f)
        for data in json_data:
            img_path = os.path.join(img_root, data['path'])
            # print(os.path.exists(img_path))
            img = cv2.imread(img_path)
            data['size'] = {'height': img.shape[0], 'width': img.shape[1], 'channels': img.shape[2]}
        with open(new_json_path, 'w') as f:
            json.dump(json_data, f, indent=2, ensure_ascii=False)
```

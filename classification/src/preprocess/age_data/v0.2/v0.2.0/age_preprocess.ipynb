```python
import os
import cv2
from tqdm import tqdm

def find_broken_images(folder_path):
    broken_files = []
    file_paths = []

    for root, _, files in os.walk(folder_path):
        for file in files:
            file_paths.append(os.path.join(root, file))
    for file_path in tqdm(file_paths):
        try:
            img = cv2.imread(file_path)
            if img is None:
                broken_files.append(file_path)
        except Exception as e:
            broken_files.append(file_path)
    return broken_files

# 사용 예시
folder = "/home/work/hocheol_dir/workspace/datasets/v0.2_data/age/114_data/data"
broken = find_broken_images(folder)
print(f"깨진 이미지 파일 수: {len(broken)}")
for f in broken:
    print(f)
```

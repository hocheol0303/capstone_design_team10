유저별 사람이 평가한 annotation 정리

```python
import json
import os
```

```python
tr_root_dir = '/home/work/hocheol_dir/workspace/data/annotations/028_data/Training/02.라벨링데이터/1. 디지털카메라'
val_root_dir = '/home/work/hocheol_dir/workspace/data/annotations/028_data/Validation/02.라벨링데이터/1. 디지털카메라'
id_paths = []
# 각 폴더 0~8번의 annotation 모으기
for id_dir in sorted(os.listdir(tr_root_dir)):
    id_paths.append(os.path.join(tr_root_dir, id_dir))
for id_dir in sorted(os.listdir(val_root_dir)):
    id_paths.append(os.path.join(val_root_dir, id_dir))

id_paths.sort(key=lambda x:x.split('/')[-1])
```

```python
tails = []
for i in range(9):
    tails.append(f'F_{i:>02}.json')

tails
```

```python
json_data = []
for root_dir in id_paths:
    lst = sorted(os.listdir(root_dir))
    id = root_dir.split('/')[-1]
    # print(os.listdir(root_dir))
    # print(id)
    json_data.append({'id':id})
    for item in tails:
        json_path = os.path.join(root_dir, f'{id}_01_{item}')
        # print(json_path)
        with open(json_path, 'r') as f:
            annots = json.load(f)['annotations']
            for k, v in annots.items():
                json_data[-1][k] = v
```

```python
with open('/home/work/hocheol_dir/workspace/data/annotations/user_info.json', 'w') as f:
    json.dump(json_data, f, indent=4, ensure_ascii=False)
```

```python
import json
import pandas as pd

# JSON 파일 경로 설정
json_path = '/home/work/hocheol_dir/workspace/data/annotations/user_info.json'  # 본인 경로에 맞게 수정하세요
csv_path = '/home/work/hocheol_dir/workspace/data/annotations/user_info_converted.csv'

# JSON 파일 읽기
with open(json_path, 'r') as f:
    data = json.load(f)

# 데이터 처리
processed_data = []
for item in data:
    acne_info = item.get('acne')
    
    if acne_info is None:
        acne_count = 0
        acne_types = ''
    else:
        acne_count = len(acne_info)
        acne_types = ','.join(acne['name'] for acne in acne_info)
    
    # acne 제외하고 다른 값 복사
    row = {k: v for k, v in item.items() if k != 'acne'}
    row['acne_count'] = acne_count
    row['acne_types'] = acne_types
    
    processed_data.append(row)

# pandas DataFrame으로 변환
df = pd.DataFrame(processed_data)

# CSV 파일로 저장
df.to_csv(csv_path, index=False)

print(f"CSV 파일 저장 완료: {csv_path}")
```

```python
import json
import os

root_dir_6ch = '/home/work/hocheol_dir/workspace/data/0512_merged_data/6ch_annot'

with open(os.path.join(root_dir_6ch, 'test.json'), 'r') as f:
    test_json_6ch = json.load(f)
with open(os.path.join(root_dir_6ch, 'train.json'), 'r') as f:
    train_json_6ch = json.load(f)
with open(os.path.join(root_dir_6ch, 'val.json'), 'r') as f:
    val_json_6ch = json.load(f)
```

```python
root_dir_3ch = '/home/work/hocheol_dir/workspace/data/0512_merged_data/3ch_annot'

test_json_3ch = []
train_json_3ch = []
val_json_3ch = []

for data in test_json_6ch:
    test_json_3ch.append({
        'id': data['id'],
        'cheek_path': data['left_cheek_path'],
        'label': data['label']
    })
    test_json_3ch.append({
        'id': data['id'],
        'cheek_path': data['right_cheek_path'],
        'label': data['label']
    })

for data in train_json_6ch:
    train_json_3ch.append({
        'id': data['id'],
        'cheek_path': data['left_cheek_path'],
        'label': data['label']
    })
    train_json_3ch.append({
        'id': data['id'],
        'cheek_path': data['right_cheek_path'],
        'label': data['label']
    })

for data in val_json_6ch:
    val_json_3ch.append({
        'id': data['id'],
        'cheek_path': data['left_cheek_path'],
        'label': data['label']
    })
    val_json_3ch.append({
        'id': data['id'],
        'cheek_path': data['right_cheek_path'],
        'label': data['label']
    })
```

```python
with open(os.path.join(root_dir_3ch, 'test.json'), 'w') as f:
    json.dump(test_json_3ch, f, indent=2, ensure_ascii=False)
with open(os.path.join(root_dir_3ch, 'train.json'), 'w') as f:
    json.dump(train_json_3ch, f, indent=2, ensure_ascii=False)
with open(os.path.join(root_dir_3ch, 'val.json'), 'w') as f:
    json.dump(val_json_3ch, f, indent=2, ensure_ascii=False)
```

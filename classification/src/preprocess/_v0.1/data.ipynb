```python
import json
import os
merged_root = '/home/work/hocheol_dir/workspace/data/merged_data'
tr_path = os.path.join(merged_root, 'merged_annot_6ch/train.json')
val_path = os.path.join(merged_root, 'merged_annot_6ch/val.json')
test_path = os.path.join(merged_root, 'merged_annot_6ch/test.json')

tr_json = None
val_json = None
test_json = None

total_json = []
with open(tr_path, 'r') as f:
    tr_json = json.load(f)
with open(val_path, 'r') as f:
    val_json = json.load(f)
with open(test_path, 'r') as f:
    test_json = json.load(f)

for data in [tr_json, val_json, test_json]:
    total_json.extend(data)
```

```python
import pandas as pd
df = pd.read_csv('/home/work/hocheol_dir/workspace/data/034_cropped_data/newlabel_250429.csv')
```

```python
import os
tr_list = os.listdir('../data/034_cropped_data/1-2.Training/')
val_list = os.listdir('../data/034_cropped_data/2-2.Validation/')

tr_set = set(tr_list)
val_set = set(val_list)
```

```python
# 6번 label은 빼기
# path : root는 줄거라서 train dir, val dir에서 시작하는 경로 저장

tr_data = {}
val_data = {}
class6_data = {}
class6 = 0

for item, label in df.values:
    if label == 6:
        file_name = item
        which_cheek = None
        path = None
        id = file_name.split('_')[2]
        id = f'{id}_19' if '_CRS_19' in file_name else f'{id}_46' if '_CRS_46' in file_name else f'{id}_outlier'

        if file_name.startswith('left_cheek'):
            which_cheek = 'left_cheek'
        else: # right_cheek으로 시작
            which_cheek = 'right_cheek'

        if id.split('_')[0] in tr_set:
            path = os.path.join(f'1-2.Training/{id.split("_")[0]}/{file_name}')
            if id not in class6_data.keys():
                class6_data[id] = {
                    'id' : id,
                    'left_cheek_path' : '',
                    'right_cheek_path' : '',
                    'label' : label
                }
        if id.split('_')[0] in val_set:
            path = os.path.join(f'2-2.Validation/{id.split("_")[0]}/{file_name}')
            if id not in class6_data.keys():
                class6_data[id] = {
                    'id' : id,
                    'left_cheek_path' : '',
                    'right_cheek_path' : '',
                    'label' : label
                }
        
        class6_data[id][f'{which_cheek}_path'] = path
    else:
        file_name = item
        which_cheek = None
        path = None
        id = file_name.split('_')[2]
        id = f'{id}_19' if '_CRS_19' in file_name else f'{id}_46' if '_CRS_46' in file_name else f'{id}_outlier'

        if file_name.startswith('left_cheek'):
            which_cheek = 'left_cheek'
        else: # right_cheek으로 시작
            which_cheek = 'right_cheek'

        if id.split('_')[0] in tr_set:
            path = os.path.join(f'1-2.Training/{id.split("_")[0]}/{file_name}')
            if id not in tr_data.keys():
                tr_data[id] = {
                    'id' : id,
                    'left_cheek_path' : '',
                    'right_cheek_path' : '',
                    'label' : label
                }

                if which_cheek == 'left_cheek':
                    tr_data[id]['left_cheek_path'] = path
                    tr_data[id]['right_cheek_path'] = path.replace('left_cheek', 'right_cheek')
                else:
                    tr_data[id]['right_cheek_path'] = path
                    tr_data[id]['left_cheek_path'] = path.replace('right_cheek', 'left_cheek')


        if id.split('_')[0] in val_set:
            path = os.path.join(f'2-2.Validation/{id.split("_")[0]}/{file_name}')
            if id not in val_data.keys():
                val_data[id] = {
                    'id' : id,
                    'left_cheek_path' : '',
                    'right_cheek_path' : '',
                    'label' : label
                }

                if which_cheek == 'left_cheek':
                    val_data[id]['left_cheek_path'] = path
                    val_data[id]['right_cheek_path'] = path.replace('left_cheek', 'right_cheek')
                else:
                    val_data[id]['right_cheek_path'] = path
                    val_data[id]['left_cheek_path'] = path.replace('right_cheek', 'left_cheek')
```

```python
tr_data_list = []
val_data_list = []

for key, value in tr_data.items():
    tr_data_list.append(value)
tr_data_list.sort(key=lambda x:x['id'])

for key, value in val_data.items():
    val_data_list.append(value)
val_data_list.sort(key=lambda x:x['id'])
```

```python
for data in tr_data_list:
    if data['left_cheek_path'] == '':
        data['left_cheek_path'] = data['right_cheek_path'].replace('right', 'left')
    elif data['right_cheek_path'] == '':
        data['right_cheek_path'] = data['left_cheek_path'].replace('left', 'right')
for data in val_data_list:
    if data['left_cheek_path'] == '':
        data['left_cheek_path'] = data['right_cheek_path'].replace('right', 'left')
    elif data['right_cheek_path'] == '':
        data['right_cheek_path'] = data['left_cheek_path'].replace('left', 'right')
```

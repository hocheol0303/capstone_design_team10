```python
import pandas as pd
import json
import os

with open('/home/work/hocheol_dir/workspace/data/0512_merged_data/6ch_annot/test.json', 'r') as f:
    merged_test = pd.DataFrame(json.load(f))
with open('/home/work/hocheol_dir/workspace/data/0512_merged_data/6ch_annot/train.json', 'r') as f:
    merged_train = pd.DataFrame(json.load(f))
with open('/home/work/hocheol_dir/workspace/data/0512_merged_data/6ch_annot/val.json', 'r') as f:
    merged_val = pd.DataFrame(json.load(f))

with open('/home/work/hocheol_dir/workspace/data/028_data/028_cropped_data_mp/with_forehead/test.json', 'r') as f:
    forehead_test = pd.DataFrame(json.load(f)).loc[:, ['id', 'forehead_path', 'age']]
with open('/home/work/hocheol_dir/workspace/data/028_data/028_cropped_data_mp/with_forehead/train.json', 'r') as f:
    forehead_train = pd.DataFrame(json.load(f)).loc[:, ['id', 'forehead_path', 'age']]
with open('/home/work/hocheol_dir/workspace/data/028_data/028_cropped_data_mp/with_forehead/val.json', 'r') as f:
    forehead_val = pd.DataFrame(json.load(f)).loc[:, ['id', 'forehead_path', 'age']]

with open('/home/work/hocheol_dir/workspace/data/034_cropped_data/0512_034_dataset/with_forehead/test.json', 'r') as f:
    concat_test = pd.concat([forehead_test, pd.DataFrame(json.load(f)).loc[:, ['id', 'forehead_path', 'age']]])
with open('/home/work/hocheol_dir/workspace/data/034_cropped_data/0512_034_dataset/with_forehead/train.json', 'r') as f:
    concat_train = pd.concat([forehead_train, pd.DataFrame(json.load(f)).loc[:, ['id', 'forehead_path', 'age']]])
with open('/home/work/hocheol_dir/workspace/data/034_cropped_data/0512_034_dataset/with_forehead/val.json', 'r') as f:
    concat_val = pd.concat([forehead_val, pd.DataFrame(json.load(f)).loc[:, ['id', 'forehead_path', 'age']]])

forehead_all = pd.concat([concat_train, concat_val, concat_test], axis=0)
```

```python
merged_forehead_test = pd.merge(merged_test, forehead_all, how='left', on='id')
merged_forehead_train = pd.merge(merged_train, forehead_all, how='left', on='id')
merged_forehead_val = pd.merge(merged_val, forehead_all, how='left', on='id')
```

```python
merged_forehead_test['age'].isna().sum(), merged_forehead_train['age'].isna().sum(), merged_forehead_val['age'].isna().sum()
merged_forehead_train[merged_forehead_train['age'].isna() != True]['age'].isna().sum()
merged_forehead_train = merged_forehead_train[merged_forehead_train['age'].isna() != True]
```

```python
merged_forehead_test['age'].isna().sum(), merged_forehead_train['age'].isna().sum(), merged_forehead_val['age'].isna().sum()
```

```python
merged_forehead_test.drop(columns=['pigment_count'], inplace=True)
merged_forehead_train.drop(columns=['pigment_count'], inplace=True)
merged_forehead_val.drop(columns=['pigment_count'], inplace=True)
```

```python
with open('/home/work/hocheol_dir/workspace/data/0518_merged_data/test.json', 'w') as f:
    json.dump(merged_forehead_test.to_dict(orient='records'), f, indent=2, ensure_ascii=False)
with open('/home/work/hocheol_dir/workspace/data/0518_merged_data/train.json', 'w') as f:
    json.dump(merged_forehead_train.to_dict(orient='records'), f, indent=2, ensure_ascii=False)
with open('/home/work/hocheol_dir/workspace/data/0518_merged_data/val.json', 'w') as f:
    json.dump(merged_forehead_val.to_dict(orient='records'), f, indent=2, ensure_ascii=False)
```

```python
cols = df.columns[1:-1]
valid_df = df[df.apply(lambda row: any(row[cols] != -1), axis=1)]
valid_df[cols] = valid_df[cols].astype(int)
valid_df.rename(columns={'periroal': 'perioral'}, inplace=True)
valid_df
```

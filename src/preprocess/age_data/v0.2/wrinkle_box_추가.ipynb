```python
import mediapipe as mp
import os
import json
from tqdm import tqdm
import sys
sys.path.append('/home/work/hocheol_dir/workspace/preprocess2/age_data/v0.2')
from wrinkle_box import combined_process

import matplotlib.pyplot as plt
import matplotlib.patches as patches
```

```python
paths_json = '/home/work/hocheol_dir/workspace/datasets2/age_data/v0.2/age_label_paths.json'
img_root = '/home/work/hocheol_dir/workspace/datasets2/age_data/v0.2/v0.2_data'

face_mesh = mp.solutions.face_mesh.FaceMesh(static_image_mode=True, max_num_faces=1, refine_landmarks=True)

with open(paths_json, 'r') as f:
    paths = json.load(f)
for i, path in enumerate(paths):
    print(f"{i+1} / {len(paths)}: {path}")
    for step in ['train', 'val', 'test']:
        print(f"  - {step}")
        json_path = os.path.join(path, f"age_{step}.json")
        new_json_path = os.path.join(path, 'wrinkle_boxes', f"age_{step}.json")
        # os.makedirs(os.path.dirname(new_json_path), exist_ok=True)
        with open(json_path, 'r') as f:
            json_data = json.load(f)
        # for data in json_data:
        for data in tqdm(json_data):
            img_path = os.path.join(img_root, data['path'])
            # print(os.path.exists(img_path), img_path)
            origin_img_path = img_path.replace('v0.2/v0.2_data', 'origin_data')
            # print(os.path.exists(origin_img_path), origin_img_path)
            #################### 이까지 됨

            # print(os.path.exists(img_path))
            try:
                processed_image, boxes, cropped_image, adjusted_boxes = combined_process(origin_img_path, face_mesh)
                h, w, c = processed_image.shape
                for sector, box in adjusted_boxes.items():
                    y_min, x_min, y_max, x_max = box['y_min'], box['x_min'], box['y_max'], box['x_max'],

                    y_min = 0 if y_min < 0 else y_min
                    x_min = 0 if x_min < 0 else x_min
                    y_max = h if y_max > h else y_max
                    x_max = w if x_max > w else x_max

                    adjusted_boxes[sector] = {"y_min": int(y_min), "x_min": int(x_min), "y_max": int(y_max), "x_max": int(x_max)}

                data['boxes'] = adjusted_boxes
            except Exception as e:
                data['boxes'] = False
                print(f"Error processing image {img_path}: {e}")
                continue

        with open(new_json_path, 'w') as f:
            json.dump(json_data, f, indent=2, ensure_ascii=False)
```

```python
import json
import os

with open('/home/work/hocheol_dir/workspace/datasets2/pigment_data/v0.3/pig_label_paths.json', 'r') as f:
    label_paths = json.load(f)
# label_paths

json_data = [] 

for e in label_paths:
    for step in ['train', 'val', 'test']:
        file_path = os.path.join(e, f"pig_{step}.json")
        with open(file_path, 'r') as f:
            json_data.extend(json.load(f))
len(json_data)
```

```python
from tqdm import tqdm
import os
image_root = '/home/work/hocheol_dir/workspace/datasets/v0.2_data/pigment'
for data in tqdm(json_data):
    left_path = data.get('left_path', None)
    right_path = data.get('right_path', None)

    if left_path:
        if not os.path.exists(os.path.join(image_root, left_path)):
            print(data)
            break
    if right_path:
        if not os.path.exists(os.path.join(image_root, right_path)):
            print(data)
            break
# 전부 있음
```

```python
# v0.3 경로로 바꿔서 확인 : v0.2 경로에서 배경만 지워서 v0.3 만들었으니 경로만 바꿔서 돌렸을 때 다 있어야 정상
import os
from tqdm import tqdm

image_root = '/home/work/hocheol_dir/workspace/datasets2/pigment_data/v0.3/v0.3_data'

for data in tqdm(json_data):
    left_path = data.get('left_path', None)
    right_path = data.get('right_path', None)

    if left_path:
        left_path = os.path.join(image_root, left_path)
        if not os.path.exists(left_path):
            print(data)
            break
    if right_path:
        right_path = os.path.join(image_root, right_path)
        if not os.path.exists(right_path):
            print(data)
            break
```

```python
new_data_dir = '/home/work/hocheol_dir/workspace/datasets2/pigment_data/v0.3/v0.3_data'

for path in label_paths:
    suffix = path[path.find('/pigment')+len('/pigment'):]
    new_label_path = new_data_dir + suffix
    for step in ['train', 'val', 'test']:
        json_data = []
        with open(os.path.join(path, f"pig_{step}.json"), 'r') as f:
            json_data = json.load(f)
        
        os.makedirs(new_label_path, exist_ok=True)
        with open(os.path.join(new_label_path, f"pig_{step}.json"), 'w') as f:
            json.dump(json_data, f, indent=2, ensure_ascii=False)
```

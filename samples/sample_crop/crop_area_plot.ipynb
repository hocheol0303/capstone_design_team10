```python
from preprocess.wrinkle.wrinkle_preprocess import wrinkle_preprocess
from preprocess.remove_non_skin import remove_non_skin
from utils.Mediapipe import get_landmarks, MediaPipe, get_mp_image
```

```python
face_landmarker = MediaPipe()
mp_image = get_mp_image(face_landmarker, "/home/hocheol/inskin_ai/no_track/datasets/origin_data-simple/034_data/0000/0000_CRS_19_01.jpg")
landmarks, np_image = get_landmarks(face_landmarker, mp_image)
np_image = remove_non_skin(np_image, landmarks)
results = wrinkle_preprocess(np_image, landmarks)
```

```python
import os
import cv2

root_dir = '/home/hocheol/inskin_ai/no_track/tmp/sample_crop/wrinkle'

os.makedirs(root_dir, exist_ok=True)

np_image_rgb = cv2.cvtColor(np_image, cv2.COLOR_BGR2RGB)
cv2.imwrite(os.path.join(root_dir, 'remove_non_skin.jpg'), np_image_rgb)

for i, bgr in enumerate(results):
    rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    cv2.imwrite(os.path.join(root_dir, f'{i}.jpg'), rgb)
```

```python

from preprocess.pigment_data.crop_mp import crop_cheeks_from_np_image

root_dir = '/home/hocheol/inskin_ai/no_track/tmp/sample_crop/cheeks'
os.makedirs(root_dir, exist_ok=True)

cv2.imwrite(os.path.join(root_dir, 'remove_non_skin.jpg'), np_image_rgb)
cropped_cheeks = crop_cheeks_from_np_image(np_image, landmarks)
for i, bgr in enumerate(cropped_cheeks):
    rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    cv2.imwrite(os.path.join(root_dir, f'cheek_{i}.jpg'), rgb)
```
